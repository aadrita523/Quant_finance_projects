### Splitting dataset

```python
split_cutoff = pd.Timestamp('2014-01-01')
train_credit = credit_risk[credit_risk['issue_date'] < split_cutoff].copy()
test_credit = credit_risk[credit_risk['issue_date']>= split_cutoff].copy()
```

```python
X_train = train_credit.drop(columns = ['good_bad', 'issue_date'])
y_train = train_credit['good_bad']
```

```python
X_test = test_credit.drop(columns = ['good_bad', 'issue_date'])
y_test = test_credit['good_bad']
```

```python
cols_to_drop = ['issue_d','term','emp_length','emp_length_int'] 
X_train = X_train.drop(columns=cols_to_drop, errors='ignore')
X_test = X_test.drop(columns=cols_to_drop, errors='ignore')
```

#Categorical data

```python
cat_cols = ['home_ownership', 'verification_status', 'purpose']


X_train_enc = pd.get_dummies(X_train, columns=cat_cols, drop_first=False, dtype=int)
X_test_enc = pd.get_dummies(X_test, columns=cat_cols, drop_first=False, dtype=int)



X_test_enc = X_test_enc.reindex(columns=X_train_enc.columns, fill_value=0)

X_train_enc.shape
X_test_enc.shape
```

#model fitting

```python
import xgboost as xgb
from sklearn.metrics import roc_auc_score
```

```python
xgb_model = xgb.XGBClassifier(
    n_estimators=300,        # Number of trees
    learning_rate=0.05,      # Step size (smaller = steadier learning)
    max_depth=4,             # Keep trees shallow to prevent overfitting
    subsample=0.8,           # Use 80% of rows per tree
    colsample_bytree=0.8,    # Use 80% of features per tree
    eval_metric='auc',
    random_state=42,
    n_jobs=-1
)


xgb_model.fit(X_train_enc, y_train)

```

```python
pd_train = 1 - xgb_model.predict_proba(X_train_enc)[:, 1]
pd_test = 1 - xgb_model.predict_proba(X_test_enc)[:, 1]

# 4. Quick AUC & Gini check (using 1 - y so 1 = Default)
train_auc = roc_auc_score(1 - y_train, pd_train)
test_auc = roc_auc_score(1 - y_test, pd_test)

print(f"Train AUC: {train_auc:.4f} | Train Gini: {2*train_auc - 1:.4f}")
print(f"Test AUC  : {test_auc:.4f} | Test Gini  : {2*test_auc - 1:.4f}")
```

```python
pd_train_xgb = 1 - xgb_model.predict_proba(X_train_enc)[:, 1]
pd_test_xgb  = 1 - xgb_model.predict_proba(X_test_enc)[:, 1]

bins = np.quantile(pd_train_xgb, np.linspace(0, 1, 11))
bins[0], bins[-1] = -np.inf, np.inf

train_pct = np.clip(np.histogram(pd_train_xgb, bins=bins)[0] / len(pd_train_xgb), 1e-6, None)
test_pct  = np.clip(np.histogram(pd_test_xgb,  bins=bins)[0] / len(pd_test_xgb),  1e-6, None)
```

```python
psi_xgb = np.sum((test_pct - train_pct) * np.log(test_pct / train_pct))
print(f"XGBoost Score PSI (Train vs Test): {psi_xgb:.4f}")
```

```python
pd_train_lreg = 1 - logistic_reg_model.predict_proba(X_train_new)[:, 1]
pd_test_lreg = 1 - logistic_reg_model.predict_proba(X_test_new)[:, 1]
```

```python
train_auc_lreg = roc_auc_score(1 - y_train, pd_train_lreg)
test_auc_lreg = roc_auc_score(1 - y_test, pd_test_lreg)

print(f"LR Train AUC: {train_auc_lreg:.4f} | Gini: {2*train_auc_lreg - 1:.4f}")
print(f"LR Test AUC  : {test_auc_lreg:.4f} | Gini: {2*test_auc_lreg - 1:.4f}")
```

#calculating KS

```python
p_good_test_lreg = logistic_reg_model.predict_proba(X_test_new)[:, 1]
pd_test_lreg = 1 - p_good_test_lreg
y_hat_test_proba = p_good_test_lreg
```

```python

credit_risk_actual_predicted_probs = pd.DataFrame({
    'y_test': np.asarray(y_test),
    'y_hat_test_proba': np.asarray(p_good_test_lreg)
})
```

```python
credit_risk_actual_predicted_probs =credit_risk_actual_predicted_probs.sort_values('y_hat_test_proba').reset_index(drop=True)
```

```python

credit_risk_actual_predicted_probs['Cumulative N Population'] = credit_risk_actual_predicted_probs.index + 1
credit_risk_actual_predicted_probs['Cumulative N Good'] = credit_risk_actual_predicted_probs['y_test'].cumsum()
credit_risk_actual_predicted_probs['Cumulative N Bad'] = credit_risk_actual_predicted_probs['Cumulative N Population']- credit_risk_actual_predicted_probs['Cumulative N Good']

tot_good = credit_risk_actual_predicted_probs['y_test'].sum()
tot_bad = len(credit_risk_actual_predicted_probs) - tot_good

credit_risk_actual_predicted_probs['Cumulative Perc Good'] = (
   credit_risk_actual_predicted_probs['Cumulative N Good'] / tot_good
)
credit_risk_actual_predicted_probs['Cumulative Perc Bad'] = (
    credit_risk_actual_predicted_probs['Cumulative N Bad'] / tot_bad
)
```

```python
KS = (credit_risk_actual_predicted_probs['Cumulative Perc Bad']- credit_risk_actual_predicted_probs['Cumulative Perc Good']).max()
print(f"KS statistics: {KS:.3f} ({KS*100:.2f})")
```
